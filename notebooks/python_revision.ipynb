## Args Vs Kwargs

```python
# write python functions to add number using args
# add_numbers(1, 2, 3) should return 6
def add_numbers(*args):
    total = 0
    for num in args:
        total += num
    return total

add_numbers(1,2,3)   
```

```python
def add_numbers_without_args(a,b,c):
    return a + b + c
```

```python
add_numbers_without_args(1,2,3)
```

```python
# create a funciton to print student details using kwargs(name,age,grade)
def student_details(**kwargs):
    print(f"Name:{kwargs["name"]}")
    print(f"Age: {kwargs["age"]}")
    print(f"Grade: {kwargs["grade"]}")
```

```python
student_details(name="John",age=20,grade="A")
```

```python
def student(**kwargs):
    print(kwargs)
```

```python
student(name="john",age=20,grade="A")
```
